```python
#read all pages from a pdf

from PyPDF2 import PdfReader
reader= PdfReader("cf.pdf")
for page_number, page in enumerate (reader.pages, start=1):

  print(f"\n--- Page {page_number}---")

  text= page.extract_text()
  print(text)
```

```python
#reading a particular page
from PyPDF2 import PdfReader
reader= PdfReader("cf.pdf")
page=reader.pages[2]
print(page.extract_text())
```

```python
#search for a word in a pdf
from PyPDF2 import PdfReader
reader = PdfReader("cf.pdf")
search_word="Coffee"
for page_number, page in enumerate(reader.pages,start=1):
  text=page.extract_text()
  if search_word.lower() in text.lower():
    print(f"'{search_word}' found on page {page_number}")
```

```python
# count words in pdf
from PyPDF2 import PdfReader
reader= PdfReader("cf.pdf")
word_count=0
for page in reader.pages:
  text = page.extract_text()
  if text:
    word_count += len(text.split())
print(f"Total words:", {word_count})
```

```python
#copy the first page into another PDF
from PyPDF2 import PdfReader, PdfWriter

#read original pdf
reader= PdfReader("cf.pdf")

#create new pdf
writer=PdfWriter()

#copy first page
writer.add_page(reader.pages[0])

#save as another pdf
with open("first_page.pdf","wb") as output:
  writer.write(output)
```

```python
#removing stopwords and punctuations
doc= nlp(text)
result=[]

for token in doc:
  if not token.is_stop and not token.is_punct:
    result.append(token.text)

print(" ".join(result))
```

```python
#POS Tagging (display word, POS and explaination)
doc= nlp(text)

for token in doc:
  print(
      token.text,token.pos_,
      spacy.explain(token.pos_)
  )
```

```python
#POS Tagging (extract adjectives)
doc= nlp(text)

for token in doc:
  if token.pos_ == "ADJ":
    print(token.text)
```

```python
#NER(locations)
doc = nlp(text)
for ent in doc.ents:
    print(ent.text, ent.label_)
```
