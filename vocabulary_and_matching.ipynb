# Rule-Based Matching
spaCy offers a rule-matching tool called `Matcher` that allows you to build a library of token patterns, then match those patterns against a `Doc` object to return a list of found matches. You can match on any part of a token, including its text and annotations, and you can add multiple patterns to the same matcher.

```python
import spacy
nlp = spacy.load('en_core_web_sm')
```

```python
from spacy.matcher import Matcher
matcher = Matcher(nlp.vocab)
```

```python
pattern1 = [{'LOWER': 'machinelearning'}]
pattern2 = [{'LOWER': 'machine'}, {'LOWER': 'learning'}]
pattern3 = [{'LOWER': 'machine'}, {'IS_PUNCT': True}, {'LOWER': 'learning'}]

matcher.add('MachineLearning', [pattern1, pattern2, pattern3])
```

```python
doc = nlp('The Machine Learning field continues to grow as demand '
          'for machinelearning skills increases. Machine-learning courses '
          'are becoming more popular every year.')

found_matches = matcher(doc)
print(found_matches)
```

```python
for match_id, start, end in found_matches:
    string_id = nlp.vocab.strings[match_id]  # get the string representation
    span = doc[start:end]                    # get the matched span
    print(match_id, string_id, start, end, span.text)
```

# Phrase Matcher

```python
from spacy.matcher import PhraseMatcher

# attr='LOWER' makes the matching case-insensitive
matcher = PhraseMatcher(nlp.vocab, attr='LOWER')
```

```python
text = ('Artificial intelligence is changing many industries. '
        'Deep learning is a branch of machine learning that uses neural networks. '
        'Many students now study machine learning and deep learning '
        'to build a career in artificial intelligence.')

doc3 = nlp(text)
```

```python
# create a list of phrases to match
phrase_list = ['artificial intelligence', 'machine learning', 'deep learning', 'neural networks']

# convert each phrase to a Doc object
phrase_patterns = [nlp.make_doc(phrase) for phrase in phrase_list]

# pass the Doc objects to the matcher as a list
matcher.add('AITerms', phrase_patterns)

# build a list of matches as (match_id, start, end)
matches = matcher(doc3)
matches
```

```python
for match_id, start, end in matches:
    print(start, end, doc3[start:end].text)
```
