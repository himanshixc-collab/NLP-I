```python
## 3. Tokenization with Token Index
# Each token has an index indicating its position in the document.
# token.i gives the position/index of the token.

import spacy
nlp = spacy.load("en_core_web_sm")
text = "Students love to paint"
doc = nlp(text)

for token in doc:
    print(token.i, token.text)
```

```python
#tokenization with token attributes
#spacy provides several useful attrivutes for each token
import spacy
nlp= spacy.load("en_core_web_sm")
text = "We should learn French together."
doc= nlp(text)
for token in doc:
  print(token.text,
        token.is_alpha,
        token.is_punct,
        token.is_space)
```

alpha - check whether token contains alphabetic characters
punct- checks whether

```python
#tokenization of email address

import spacy
nlp = spacy.load("en_core_web_sm")
text= "for more information contact: manshi@yahoo.com or tanisha@gmail.com"
doc= nlp(text)

for token in doc:
  print(token.text)
```

```python
#tokenization of contractions

import spacy
nlp= spacy.load("en_core_web_sm")
text= "It wasn't present"

doc= nlp(text)

for token in doc:
  print(token.text)
```

```python
#tokenization of hyphenated words

import spacy
nlp= spacy.load("en_core_web_sm")
text= "She is a well-known author."
doc= nlp(text)

for token in doc:
  print(token.text)
```

```python
#removing punctuation after tokenizaton
import spacy
nlp= spacy.load("en_core_web_sm")
text= "Students are dancing, singing and enjoying!!"
doc= nlp(text)

for token in doc:
  if not token.is_punct:
    print(token.text)
```

```python
#getting only alpj=alphabetic token
#we can use is_alpha to select only alphabetic toens

import spacy
nlp= spacy.load("en_core_web_sm")
text= "Harry potter 2 is interesting!!"
doc= nlp(text)

for token in doc:
  if token.is_alpha:
    print(token.text)
```
