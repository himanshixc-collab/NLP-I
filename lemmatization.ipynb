```python
#Removing punctuation and lemmatization
import spacy
nlp = spacy.load("en_core_web_sm")

text = "Researchers, after reviewing the evidence, revised their conclusions!"
doc = nlp(text)
words = []

for token in doc:
    if not token.is_punct:
        words.append(token.lemma_)

print(words)
```

```python
#Lemmatization with stopword removal
import spacy
nlp = spacy.load("en_core_web_sm")

text = "The research team was carefully comparing the experimental results."
doc = nlp(text)
words = []

for token in doc:
    if not token.is_stop and not token.is_punct:
        words.append(token.lemma_)

print(words)
```

```python
#Create a lemmatization function
import spacy
nlp = spacy.load("en_core_web_sm")

def lemmatize_text(text):
    doc = nlp(text)
    result = []

    for token in doc:
        if not token.is_stop and not token.is_punct:
            result.append(token.lemma_)

    return " ".join(result)

text = "Several researchers were investigating unusual linguistic patterns."
result = lemmatize_text(text)
print(result)
```

```python
#basic NLTK lemmatization

from nltk.stem import WordNetLemmatizer
lemmatizer = WordNetLemmatizer()

words = ["analysts", "laboratories", "hypotheses", "phenomena"]
for word in words:
    print(word, "->", lemmatizer.lemmatize(word))
```

```python
#lemmatizing verbs
#By default, NLTK assumes a word is a noun.
#Therefore, specify the POS when lemmatizing verbs.

from nltk.stem import WordNetLemmatizer
lemmatizer = WordNetLemmatizer()

words = ["investigating", "written", "drove", "sought"]

for word in words:
    print(word, "->", lemmatizer.lemmatize(word, pos="v"))
```

```python
#Lemmatizing nouns

from nltk.stem import WordNetLemmatizer
lemmatizer = WordNetLemmatizer()

words = ["geese", "mice", "indices", "criteria", "analyses"]
for word in words:
    print(word, "->", lemmatizer.lemmatize(word, pos="n"))
```

```python
# Lemmatizing adjectives

from nltk.stem import WordNetLemmatizer
lemmatizer = WordNetLemmatizer()

words = ["happier", "happiest", "narrower", "brightest"]

for word in words:
    print(word, "->", lemmatizer.lemmatize(word, pos="a"))
```

```python
#lemmatizing adverbs

from nltk.stem import WordNetLemmatizer
lemmatizer = WordNetLemmatizer()

words = ["faster", "harder", "earlier", "better"]
for word in words:
    print(word, "->", lemmatizer.lemmatize(word, pos="r"))
```

```python
#irregular verbs lemmatization

from nltk.stem import WordNetLemmatizer
lemmatizer = WordNetLemmatizer()

words = ["swam", "driven", "taught", "sought", "stolen"]
for word in words:
    print(word, "->", lemmatizer.lemmatize(word, pos="v"))
```

```python
#compare noun and verb lemmatization

#The same word can produce different results depending on its part of speech.

from nltk.stem import WordNetLemmatizer
lemmatizer = WordNetLemmatizer()

word = "saw"

print(word, "as a noun ->", lemmatizer.lemmatize(word, pos="n"))
print(word, "as a verb ->", lemmatizer.lemmatize(word, pos="v"))
```

```python
# lemmatization a list of words
from nltk.stem import WordNetLemmatizer
lemmatizer = WordNetLemmatizer()

words = [
    "analyses",
    "investigating",
    "investigated",
    "geese",
    "laboratories",
    "sought"
]

for word in words:
    if word.endswith("ing") or word.endswith("ed") or word in ["sought"]:
        lemma = lemmatizer.lemmatize(word, pos="v")
    else:
        lemma = lemmatizer.lemmatize(word, pos="n")

    print(word, "->", lemma)
```

```python
#complete nltk lemmatization with POS
import nltk

from nltk import word_tokenize
from nltk.stem import WordNetLemmatizer
from nltk.corpus import wordnet

nltk.download("punkt")
nltk.download("punkt_tab")
nltk.download("averaged_perceptron_tagger")
nltk.download("averaged_perceptron_tagger_eng")
nltk.download("wordnet")
nltk.download("omw-1.4")

lemmatizer = WordNetLemmatizer()

def get_wordnet_pos(word):
    tag = nltk.pos_tag([word])[0][1]

    if tag.startswith("J"):
        return wordnet.ADJ
    elif tag.startswith("V"):
        return wordnet.VERB
    elif tag.startswith("R"):
        return wordnet.ADV
    else:
        return wordnet.NOUN

text = "The researchers carefully examined unusual findings."
tokens = word_tokenize(text)

for token in tokens:
    if token.isalpha():
        pos = get_wordnet_pos(token)
        lemma = lemmatizer.lemmatize(token, pos=pos)
        print(token, "->", lemma)
```
