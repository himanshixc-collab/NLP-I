```python
# basic stemming
from nltk.stem import PorterStemmer
stemmer= PorterStemmer()
words=["caring","cared","cares","care"]
for word in words:
  print(word,"->", stemmer.stem(word))
```

```python
#stemming a sentence
from nltk.stem import PorterStemmer
stemmer= PorterStemmer()
text = "The students are playing basketball."
words = text.split()
for word in words:
  print(word, "->", stemmer.stem(word))
```

```python
#stemming with a loop
from nltk.stem import PorterStemmer
stemmer = PorterStemmer()

words = ["investigate", "investigated", "investigating", "investigates"]

for word in words:
    stem = stemmer.stem(word)
    print(f"{word:15} -> {stem}")
```

```python
#stemming word related to organize
from nltk.stem import PorterStemmer
stemmer = PorterStemmer()

words = [
    "organize",
    "organizes",
    "organized",
    "organizing",
    "organization",
    "organizational"
]

for word in words:
    print(word, "->", stemmer.stem(word))
```

```python
#stemming texts using regular expressions
import re
from nltk.stem import PorterStemmer
stemmer = PorterStemmer()

text = "Researchers are analyzing experimental results and documenting significant observations."
words = re.findall(r'\b[a-zA-Z]+\b', text.lower())

for word in words:
    print(word, "->", stemmer.stem(word))
```

```python
#create a stemmed sentence
from nltk.stem import PorterStemmer
stemmer = PorterStemmer()

text = "The researchers investigated samples and carefully documented their observations."

words = text.lower().split()
stemmed_words = []

for word in words:
    stemmed_words.append(stemmer.stem(word))

stemmed_text = " ".join(stemmed_words)
print("Original:")
print(text)

print("\nStemmed:")
print(stemmed_text)
```

```python
# stemming with lowercase conversion
import spacy
from nltk.stem import PorterStemmer

nlp = spacy.load("en_core_web_sm")
stemmer = PorterStemmer()

text = "Computational COMPUTATIONAL computationally computations"

doc = nlp(text)

for token in doc:
    if token.is_alpha:
        word = token.text.lower()
        print(token.text, "->", stemmer.stem(word))
```

```python
#create a stemmed sentence using spacy
import spacy
from nltk.stem import PorterStemmer

nlp = spacy.load("en_core_web_sm")
stemmer = PorterStemmer()

text = "The researchers are examining samples and documenting observations carefully."
doc = nlp(text)

stemmed_words = []

for token in doc:
    if token.is_alpha:
        stemmed_words.append(stemmer.stem(token.text.lower()))

stemmed_text = " ".join(stemmed_words)
print("Original:")
print(text)

print("\nStemmed:")
print(stemmed_text)
```
